# The court case duration model: the whole pipeline in one notebook

**IDSE (CS3006) end-to-end data-science project.** This notebook takes real Indian district-court
records from the raw download all the way to the model that is live at
[otlr.vercel.app/model](https://otlr.vercel.app/model), and explains, at every step, **what was
done, why, and what came out**.

**The purpose.** Indian district courts are known for delay. We ask: **which court cases get stuck,
where, and can you tell on the day a case is filed?** The answer is useful for planning, not for
predicting one person's case: a court registry can estimate how many new filings will still be open
in a year, and legal-aid offices can see which kinds of case to steer towards mediation.

**The short answer.** You can tell what *kind* of case will be quick. Among real disputes, filing-day
information barely predicts which ones finish fast, and almost all of a dispute's time is spent after
its first hearing.

It has two parts:

- **Part A, the data analysis pipeline:** the question, the data, the mess in it, how it was
  cleaned, two facts about the data that decide everything else, how the target was labelled
  honestly, and what the cleaned data shows, and where the time in a case goes.
- **Part B, the machine-learning pipeline:** the inputs, the time split, the baselines, the three
  model families, the model we serve and why, how it was tested on cases it never saw, what drives
  it, two side analyses (regression on days, k-means on districts), survival curves, its
  limitations, and one case scored live.

Every number is computed here from the cleaned data on disk (`data/clean/10/`, produced by
`scripts/build_dataset.py` and `scripts/label_cases.py`) or read from the reports the scripts wrote
(`reports/*.json`). Nothing is typed in by hand. Where a number is both computed here and in a
report, the two are compared and the result is printed. Slow steps (the random forests, the
permutation importance) are not redone; their results are read from the reports. All code is the
project's own `tareekh` package (`src/tareekh/`), not a re-implementation.

Data: Development Data Lab's judicial data (eCourts, filings 2010–2018), licence ODbL v1.0.

# Part A · The data analysis pipeline

## A1 · The question

**Why this question.** Court delay is a planning problem: registries must staff courts for cases that
last years, and legal-aid offices must decide which cases to steer towards settlement. Both need to
know which kinds of case get stuck and where, as early as possible.

**What.** So the model answers one measurable question:

> Given only what is known on the day a case is filed in a Tamil Nadu district court, how likely
> is it to be **disposed of within one year**?

"Disposed of" means the case ended *in that court*: a judgment, a dismissal, a settlement, a
Lok Adalat disposal, or a file closed because the accused absconded. A transfer to another court
is **not** an end; the case is still going somewhere else.

**Why a yes/no question and not "how many days".** Many cases in the data had not finished when
the data was collected, so their number of days is unknown. A yes/no question with a fixed horizon
can be answered honestly for every case that was watched for at least a year (Part A7 shows how).

```python
nat, tn = A["national"], A["tamil_nadu"]
print("Question:", META["question"])
print(f"Cases in the download (all India, filings 2010-2018): {nat['cases']:,}")
print(f"Cases in Tamil Nadu, the state we model:                {tn['cases']:,}")
print(f"\nTakeaway: one precise question, answered on {tn['cases']:,} real Tamil Nadu cases "
      f"out of {nat['cases'] / 1e6:.1f} million nationally.")
```

**What.** Development Data Lab (DDL) scraped the government's eCourts portal, removed names, and
published one row per case for every district court in India, filings 2010–2018. The download is
a Dropbox folder of CSV files inside tar archives: one case file per filing year (16 GB of CSV in
total), one file with the law and section of every case, and small "key" files that translate the

**Licence.** The licence file inside the download is ODbL v1.0 (the website says CC BY-NC-SA);

## A3 · The raw mess, before cleaning

**What.** Before fixing anything we measured what is wrong. Five problems, each shown with real
values from the data:

1. **Missing values hide as blanks and as sentinel codes.** A *sentinel* is a special value that
   stands for "unknown": here `-9998` means "unclear" and `-9999` means "missing".
2. **Impossible dates**: decisions in the year 0001 or 7201, `5000-01-01` placeholders, and
   cases "decided" before they were filed.
3. **One law, thousands of spellings.** Law names are typed by court staff:
   29,857 distinct strings for what are a few dozen laws.
4. **Court names are free text** with typos ("trail", "musif", "subardinate").
5. **Section numbers are free text** too: "138r/w142", "OVIIR1".

**Why measure first.** Every cleaning rule costs something (a wrong merge is worse than an honest
"unknown"), so each rule below reports how many rows it touched.

```python
# 1. blanks and sentinels, as a share of the raw 2017 rows
def missing(col):
    s = raw[col]
    return pd.Series({"blank": (s == "").mean(),
                      "-9998 (unclear)": s.str.startswith("-9998").mean(),
                      "-9999 (missing)": s.str.startswith("-9999").mean()})


miss = pd.DataFrame({c: missing(c) for c in raw.columns}).T
print("Columns with blanks or sentinel codes (share of 2017 Tamil Nadu rows):")
display(miss[(miss > 0).any(axis=1)].style.format("{:.1%}"))

# 2. impossible dates: the raw year typed, per date column
dates = ["date_of_decision", "date_first_list", "date_last_list", "date_next_list"]
typed = pd.DataFrame({c: {"earliest year typed": pd.to_numeric(raw[c].str[:4], errors="coerce").min(),
                          "latest year typed": pd.to_numeric(raw[c].str[:4], errors="coerce").max()}
                      for c in dates}).T.astype("Int64")
display(typed)

# 3. one law, many spellings (act_key.csv lists every raw law name in the country with its count)
acts_key = key("act_key.csv")
acts_key["count"] = pd.to_numeric(acts_key["count"])
ni = acts_key[acts_key["act_s"].str.contains(r"negotiab|n\.? ?i\.? ?act", case=False, regex=True)]
print(f"\n{len(acts_key):,} distinct raw law names in the country. "
      f"{len(ni):,} of them are spellings of the Negotiable Instruments Act, for example:")
display(ni.nlargest(10, "count")[["act_s", "count"]].rename(columns={"act_s": "raw law name"}))

# 4. court names and 5. section strings
courts = raw["judge_position"]
odd = courts[courts.str.contains("trail|musif|subardinate|mgistrate|magistarte")].unique()
print(f"{courts.nunique()} distinct court names in 2017 alone; with typos such as:", list(odd[:4]))
section_key = key("section_key.csv")
ni_sections = section_key[section_key["section_s"].str.contains(r"^138", regex=True)]["section_s"]
print(f"{len(ni_sections):,} distinct ways of writing section 138 (cheque bounce), e.g.",
      list(ni_sections.sample(6, random_state=3)))
print("\nTakeaway: nothing can be counted until law names, court names, sections, dates and "
      "sentinel codes are normalised; every rule for that is tested on these real strings.")
```

## A4 · Cleaning, with before and after counts

**What.** Each problem gets one small, tested function in `src/tareekh/` (138 tests, run with
`uv run pytest`):

- **Law names** → `acts.canonical_act`: lower-case, keep letters only, then ordered regular
  expressions from specific to general; unmatched names stay "Other" rather than being guessed.
- **Section strings** → `sections.primary_section`: keep the first section number and its letter
  suffix: "304(B)" → 304B, "138r/w142" → 138, "OVIIR1" → Order 7.
- **Court names** → `courts.court_level`: 15 court levels by ordered regular expressions
  ("Other / unclear" when nothing matches).
- **Dates** → `cases.decode`: parse; anything outside 1950–2022 is blanked and counted; "decided
  before filing" is flagged.
- **Sentinels** → `cases.decode`: `-9998` → "unclear", `-9999` → "missing".
- **52 disposition categories** → `outcomes.outcome_group`: each put in one group: decided, settled,
  dismissed, disposed, dormant, **moved** (transfer, committal, remand, referral: the case left,
  it did not end), interim (bail orders), missing.

**Why these choices.** We never "repair" a date (we cannot know the intended one) and never guess a
law; we count what we could not fix. A **moved** case is deliberately not treated as finished:
treating transfers as endings would make courts look faster than they are.

The cell below applies the real functions and prints how much each one changed.

```python
from tareekh.acts import UNKNOWN, canonical_act
from tareekh.courts import court_level
from tareekh.outcomes import outcome_group
from tareekh.sections import primary_section

lab = pd.read_parquet(CLEAN / "labelled.parquet")   # every Tamil Nadu case, decoded (scripts/label_cases.py)

# laws: 29,857 raw names -> a few dozen laws, weighted by how many case rows use each name
acts_key["law"] = acts_key["act_s"].map(canonical_act)
named = ~acts_key["law"].isin([UNKNOWN, "Other"])
rows = acts_key["count"].sum()
def share(m):
    return acts_key.loc[m, "count"].sum() / rows


law_named, law_junk, law_other = share(named), share(acts_key.law == UNKNOWN), share(acts_key.law == "Other")
check("share of law rows mapped to a named law", law_named, tn["cleaning"]["law_rows_named"])

# sections: distinct strings before and after, for four big laws
acts = pd.read_parquet(CLEAN / "acts_sections.parquet", columns=["act", "section"])
acts["law"] = acts["act"].map(acts_key.set_index("act")["act_s"]).fillna("").map(canonical_act)
raw_sec = acts["section"].map(section_key.set_index("section")["section_s"]).fillna("")
sections_table = pd.DataFrame({
    law: {"raw strings": raw_sec[acts.law == law].nunique(),
          "after primary_section()": raw_sec[acts.law == law].map(primary_section).nunique()}
    for law in ["Indian Penal Code", "Negotiable Instruments Act", "Motor Vehicles Act",
                "Code of Civil Procedure"]}).T

# courts, dates, sentinels, outcomes
lab["court_level"] = lab["judge_position"].map(court_level)
lab["group"] = lab["disp"].map(outcome_group)
before_after = pd.DataFrame([
    ["law names (all India)", f"{len(acts_key):,} raw strings",
     (f"{acts_key.loc[named, 'law'].nunique()} laws; {law_named:.1%} of rows named, "
      f"{law_junk:.1%} junk ('.', '---', 'ACT'), {law_other:.1%} left as Other")],
    ["court names (Tamil Nadu)", f"{lab['judge_position'].nunique()} raw strings",
     f"{lab['court_level'].nunique()} court levels (incl. unclear)"],
    ["decision dates", "years 0001 to 9201 typed",
     (f"{int(lab['date_of_decision_impossible'].sum()):,} blanked as impossible; "
      f"{int(lab['decided_before_filing'].sum()):,} decided before filing, set aside")],
    ["last-hearing dates", "5000-01-01 placeholders",
     f"{int(lab['date_last_list_impossible'].sum()):,} blanked"],
    ["gender", "-9998 / -9999 codes",
     (f"defendant unclear {(lab['defendant'] == 'unclear').mean():.0%}, "
      f"petitioner unclear {(lab['petitioner'] == 'unclear').mean():.0%}; kept out of the model")],
    ["dispositions", "52 categories", f"{lab['group'].nunique()} groups"],
], columns=["field", "before", "after"]).set_index("field")
display(before_after)
print("Section strings, distinct values before and after:")
display(sections_table)
groups = lab["group"].value_counts(normalize=True)
print("Outcome groups (share of all Tamil Nadu cases):")
display(groups.rename("share").to_frame().style.format("{:.1%}"))
check("share of cases in the 'moved' group", groups["moved"], tn["cleaning"]["outcome_groups"]["moved"])

fig, ax = plt.subplots(figsize=(8, 3.8))
spellings = acts_key[named].groupby("law").size().nlargest(10)
ps.hbar(ax, [s.replace(" (state liquor laws)", "") for s in spellings.index],
        list(spellings.values), fmt="{:,}")
ax.set_title("Distinct raw spellings per law, ten most-misspelt laws (all India)")
ax.set_xlabel("distinct raw spellings mapped to this law")
plt.show()

print(f"Takeaway: {len(acts_key):,} law spellings became {acts_key.loc[named, 'law'].nunique()} laws "
      f"({law_named:.1%} of rows named), {lab['judge_position'].nunique()} court names became "
      f"{lab['court_level'].nunique()} levels, {int(lab['decided_before_filing'].sum()):,} impossible "
      f"cases were set aside, and {groups['moved']:.0%} of cases are 'moved', not finished.")
```

## A5 · When was the data collected? Estimating the scrape date

**What.** The data has no "collected on" column, yet that date decides everything: a case with no
decision is only known to be *still open up to that day*. The data leaves a fingerprint: the
number of decisions recorded per month runs steadily, then collapses when scraping happened.

**How.** `cutoff.cliff` finds the **first lasting collapse**: the first month that, with the two
months after it, sits below a share of the median of the previous twelve months. We use the share
0.5 ("at least half a normal month"), because the scrape ran *during* February 2019 and a partly
scraped month would otherwise make decided cases look pending. It is computed **per district**,
with the state's value as a fallback when a small district's own dip is not a scrape.

**Why not "the last normal-looking month".** A first version did that and placed Tamil Nadu's
cutoff in July 2020: after a scrape the small trickle of re-scraped courts becomes the new normal,
so "last normal month" drifts years late. The bug became a test.

```python
from tareekh.cutoff import cliff, cutoffs, with_fallback

monthly = pd.Series(nat["monthly_decisions"])
monthly.index = pd.PeriodIndex(monthly.index, freq="M").to_timestamp()
fig, ax = plt.subplots(figsize=(8, 3.8))
ax.plot(monthly.index, monthly.values / 1e3, color=ps.SERIES[0])
normal = monthly.loc["2018"].median()
ax.axvline(pd.Timestamp("2019-02-01"), color=ps.INK_2, lw=1, ls="--")
ax.annotate("collapse: Feb 2019", (pd.Timestamp("2019-03-10"), normal / 1e3 * 0.95), fontsize=9,
            color=ps.INK_2)
ax.set_title("Decisions recorded per month, all India (thousands)")
ax.set_ylabel("decisions (thousands)")
plt.show()
print(f"A normal month in 2018: {normal:,.0f} decisions; January 2019: {monthly.loc['2019-01'].iloc[0]:,.0f}; "
      f"February: {monthly.loc['2019-02'].iloc[0]:,.0f}; March: {monthly.loc['2019-03'].iloc[0]:,.0f}.")

# the same rule, run live on Tamil Nadu's decision dates (all filing years together)
state_cutoff = cliff(lab["date_of_decision"], share=0.5)
district_cutoffs = cutoffs(lab, by=["state_code", "dist_code"], share=0.5)
district_cutoffs = with_fallback(district_cutoffs, cutoffs(lab, by="state_code", share=0.5),
                                 parent_of=lambda k: k[0])
saved = pd.read_parquet(CLEAN / "cutoffs.parquet")
print(f"\nTamil Nadu as a whole: last reliable day = {state_cutoff.date()}")
print("Per district:", district_cutoffs.dt.date.value_counts().sort_index().to_dict())
same = (district_cutoffs.reset_index(drop=True) == saved["cutoff"]).all()
print(f"Matches data/clean/10/cutoffs.parquet used for the labels: {same}")
print(f"\nTakeaway: the data was collected in January-February 2019; each district's last reliable "
      f"month is between {district_cutoffs.min().date()} and {district_cutoffs.max().date()}.")
```

## A6 · The early years are incomplete (survivorship)

**What.** Look only at the Tamil Nadu courts that already appear in 2010 and count the cases they
record per filing year. The same courts record 3.6× more cases in 2018 than in 2010, and the share
of cases decided *on the day they were filed* grows from under 1% to 18%.

**Why this matters.** Courts did not become 3.6 times busier. When a court went digital, its
pending cases were typed in, but cases that had already finished quickly mostly were not. So the
early years are missing their fast cases and look far slower than they really were. This is
*survivorship bias* (only the long-lived cases "survived" into the records). **Decision: the model
learns from 2015–2017 filings only.**

```python
lab["court"] = lab["dist_code"] + "-" + lab["court_no"]
first_year = lab.groupby("court")["filing_year"].min()
old_courts = first_year[first_year == 2010].index
in_old = lab[lab["court"].isin(old_courts)].groupby("filing_year").size()
same_day = lab.groupby("filing_year")["observed_days"].apply(lambda s: (s == 0).mean())
check("cases filed in 2018 in the courts present since 2010", in_old[2018],
      tn["survivorship"]["cases_in_those_courts"]["2018"], tol=0.5)

fig, ax = plt.subplots(figsize=(8, 3.6))
ax.bar(in_old.index, in_old.values / 1e3, color=ps.SERIES[0], width=0.7, edgecolor=ps.SURFACE, linewidth=2)
for x, v in in_old.items():
    ax.text(x, v / 1e3 + 8, f"{v / 1e3:.0f}k", ha="center", fontsize=8, color=ps.INK_2)
ax.set_title(f"Cases per filing year in the {len(old_courts)} Tamil Nadu courts present since 2010 (thousands)")
ax.set_ylabel("cases (thousands)")
ax.grid(axis="x", visible=False)
ax.set_ylim(0, in_old.max() / 1e3 * 1.15)
plt.show()

fig, ax = plt.subplots(figsize=(8, 3.2))
ax.plot(same_day.index, same_day.values, marker="o", color=ps.SERIES[1])
for x, v in same_day.items():
    ax.text(x, v + 0.008, f"{v:.1%}", ha="center", fontsize=8, color=ps.INK_2)
ax.yaxis.set_major_formatter(PercentFormatter(1.0))
ax.set_title("Share of cases decided on the day they were filed, by filing year (Tamil Nadu)")
ax.set_ylim(0, same_day.max() * 1.25)
plt.show()

print(f"Takeaway: the same {len(old_courts)} courts record {in_old[2010]:,} cases filed in 2010 and "
      f"{in_old[2018]:,} filed in 2018 (x{in_old[2018] / in_old[2010]:.1f}); same-day decisions rise from "
      f"{same_day[2010]:.1%} to {same_day[2018]:.1%}. The quick early cases were never entered, so "
      f"modelling uses 2015-2017 filings only.")
```

## A7 · Labelling "finished within a year" without guessing

**What.** `target.label` gives every case two numbers: `observed_days`, how long we watched it
(until it ended, until it moved to another court, or until its district's cutoff), and `ended`,
whether it finished in this court by the cutoff. A case still open at the cutoff is **censored**:
we know it lasted *at least* that long, not how long in total.

`target.within(…, 365)` then answers "disposed of within 365 days?" with **1** (ended by day 365),
**0** (still open after day 365) or **unknown**. It is only defined for cases filed **at least a
year before their cutoff**, so that every labelled case had the full year to finish. A case that
moved to another court before day 365 stays unknown.

**Why the eligibility rule.** A first version labelled every case it could. For 2018 filings that
meant: cases that finished fast were labelled 1, while their slow neighbours (filed a few months
before the cutoff, still open) could not be labelled at all. 2018 came out more than 90% "within a year",
which is nonsense. The cell reproduces that trap and then the honest label.

```python
from tareekh.target import within

usable = lab[lab["unusable"] == ""]
print("Cases set aside as unusable, with the reason:")
display(lab["unusable"].value_counts().drop("", errors="ignore").rename("cases").to_frame())

# the trap: label every case whose answer is "known" so far, with no eligibility rule
fast = usable["ended"] & (usable["observed_days"] <= 365)
slow = usable["observed_days"] > 365
naive = pd.Series(np.where(fast, 1.0, np.where(slow, 0.0, np.nan)), index=usable.index)
honest = within(lab, 365)
by_year = pd.DataFrame({
    "naive label: share 'within a year'": naive.groupby(usable["filing_year"]).mean(),
    "honest label: share 'within a year'": honest.groupby(lab["filing_year"]).mean(),
    "honest label: cases labelled": honest.groupby(lab["filing_year"]).count(),
}).loc[2014:2018]
display(by_year.style.format({"naive label: share 'within a year'": "{:.1%}",
                              "honest label: share 'within a year'": "{:.1%}",
                              "honest label: cases labelled": "{:,.0f}"}))
same_as_saved = honest.equals(lab["within_1y"]) or (honest.fillna(-1) == lab["within_1y"].fillna(-1)).all()
print(f"Recomputed label identical to the saved within_1y column: {same_as_saved}")
for y in (2015, 2016, 2017):
    check(f"share within a year, {y} filings", honest[lab.filing_year == y].mean(), tn["within_1y_by_year"][str(y)])

fig, ax = plt.subplots(figsize=(8, 3.4))
ax.plot(by_year.index, by_year.iloc[:, 0], marker="o", color=ps.SERIES[1], label="naive label")
ax.plot(by_year.index, by_year.iloc[:, 1], marker="o", color=ps.SERIES[0], label="honest label (full year of follow-up)")
ax.annotate("naive label", (2018, by_year.iloc[-1, 0]), xytext=(6, 0), textcoords="offset points", va="center", fontsize=9, color=ps.SERIES[1])
ax.annotate("honest label", (2018, by_year.iloc[-1, 1]), xytext=(6, 0), textcoords="offset points", va="center", fontsize=9, color=ps.SERIES[0])
ax.yaxis.set_major_formatter(PercentFormatter(1.0))
ax.set_xticks(by_year.index)
ax.set_xlim(2013.8, 2018.9)
ax.set_title("Share of filings 'disposed of within a year', naive vs honest label (Tamil Nadu)")
plt.show()

elig = honest.notna() & lab["filing_year"].between(2015, 2017)
print(f"Takeaway: the naive label says {by_year.iloc[-1, 0]:.1%} of 2018 filings finished within a year; "
      f"with full follow-up required it is {by_year.iloc[-1, 1]:.1%}. {int(elig.sum()):,} cases filed "
      f"2015-2017 carry an honest label ({honest[elig].mean():.0%} within a year).")
```

## A9 · Where the time goes in a real dispute

**What.** Sections A1–A8 show *how long* cases take. This asks *why*, as far as the data allows, for
**real disputes**: every case except petty summary trials and bail-type petitions, which are usually
over in days. Each case is split at its first hearing:

- **Waiting for the first hearing:** filing date to first hearing date.
- **After the first hearing:** first hearing to the end of the case, by Kaplan–Meier so that cases
  still running count instead of being dropped.

Two more views: the gap between a case's latest hearing and its next listed date, and the stage
recorded at the latest hearing for cases that were still open when the data was collected.

**Why it can't be a model input.** Hearing dates happen after filing, so using them to predict from
filing day would be leakage. They are used here only to describe what happens.

**Data quality.** Audited first (filings 2015–2017): 4.7% of first-hearing dates are blank and 0.3% are
before filing (both excluded here); the latest hearing date equals the decision date in 80% of finished
cases. Each case records only one gap between hearings, so the hearing count below is a rough estimate.

```python
from tareekh.difficulty import EASY_TYPES, hearing_gap, stage_days, stage_summary, stuck_at
from tareekh.glossary import PURPOSE_PLAIN

recent = usable[usable["filing_year"].between(2015, 2017)]
hard = recent[~recent["type"].isin(EASY_TYPES)]
stages = stage_days(hard)
overall = stage_summary(stages)
gap = hearing_gap(hard)
check("median days to first hearing, real disputes", overall["median_wait_first_hearing"],
      H["stages"]["contested"]["median_wait_first_hearing"])
check("median days after first hearing, real disputes", overall["median_after_first_hearing"],
      H["stages"]["contested"]["median_after_first_hearing"])
check("median gap between hearings, real disputes", gap.median(),
      H["stages"]["contested"]["median_gap_between_hearings"])

rows = [{"case type": CASE_TYPES.get(t, t.upper()).split(" (")[0],
         **stage_summary(stages.loc[part.index])}
        for t, part in hard.groupby("type") if len(part) >= 30_000]
by_stage = pd.DataFrame(rows).dropna(subset=["median_after_first_hearing"]).sort_values("median_after_first_hearing")
fig, ax = plt.subplots(figsize=(8.5, 3.8))
ax.barh(by_stage["case type"], by_stage["median_wait_first_hearing"], color=ps.SERIES[1], height=0.65, label="waiting for the first hearing")
ax.barh(by_stage["case type"], by_stage["median_after_first_hearing"], left=by_stage["median_wait_first_hearing"], color=ps.SERIES[0],
        height=0.65, label="after the first hearing", edgecolor=ps.SURFACE, linewidth=2)
ax.grid(axis="y", visible=False)
ax.legend(loc="lower right")
ax.set_xlabel("median days")
ax.set_title("Almost all the time in a dispute comes after the first hearing (TN filings 2015-2017)")
plt.show()

top = pd.DataFrame(stuck_at(hard))
top["stage"] = top["purpose"].map(PURPOSE_PLAIN).fillna(top["purpose"])
fig, ax = plt.subplots(figsize=(8.5, 4))
ps.hbar(ax, list(top["stage"]), list(top["share"]))
ax.xaxis.set_major_formatter(PercentFormatter(1.0))
ax.set_title("Stage recorded at the latest hearing, disputes still open when the data was collected")
plt.show()

after, wait, each = overall["median_after_first_hearing"], overall["median_wait_first_hearing"], gap.median()
print(f"Takeaway: a real dispute waits a median of {wait:.0f} days for its first hearing, then runs a median of {after:,.0f} days more, "
      f"so {after / (after + wait):.0%} of its time comes after the first hearing. With about {each:.0f} days between hearing dates, "
      f"that is very roughly {after / each:.0f} hearing dates. Open disputes are most often at: "
      f"{top['stage'].iloc[0].lower()} ({top['share'].iloc[0]:.0%}), {top['stage'].iloc[1].lower()} ({top['share'].iloc[1]:.0%}), "
      f"{top['stage'].iloc[2].lower()} ({top['share'].iloc[2]:.0%}).")
```

# Part B · The machine-learning pipeline

## B1 · The inputs: only what is known on filing day

**What.** `features.build` and `features.add_law` turn each case into 12 inputs, in two sets:

*Filing basics*
- `district`, `court_level`, `case_type` (the court's own type code, e.g. `os` = original civil
  suit, `cc` = calendar case, a criminal trial before a magistrate), `filing_month`;
- congestion, measured from earlier cases only: `log_pending_at_filing` (how many cases were
  already waiting in that district's courts of that level on the filing day), `clearance_last_year`
  (how many they cleared in the previous 365 days, relative to that queue) and
  `log_prev_year_filings` (filings of the same type in the district the year before). "log" means
  we use the logarithm, so 10 → 100 matters as much as 100 → 1,000.

*+ law and section* (from the acts file: what the case was filed under)
- `law` (the cleaned law), `law_section` (the 40 most common law + section pairs, e.g. "Negotiable
  Instruments Act s.138" = cheque bounce), `ipc_bailable`, `criminal` (criminal or civil),
  `n_ipc_sections`.

**Deliberately excluded.** Hearing dates and hearing purposes (they describe the case's *future*
and would leak the answer), the decision date and disposition (the answer itself), the parties'
gender (fairness: a public tool must not score people by gender), and the court number (a visitor
cannot know which numbered court will get the case).

**Leakage** means letting the model see information that would not exist at prediction time; it
makes test scores look great and the deployed model useless. The vocabularies (which case types
and sections get their own category) are decided on the training years only, for the same reason.

```python
from tareekh.features import FEATURE_SETS
from tareekh.modelling import TRAIN_YEARS, modelling_data

for name, (cats, nums) in FEATURE_SETS.items():
    print(f"{name:>18}: {len(cats) + len(nums)} inputs = categorical {cats} + numeric {nums}")

t0 = time.time()
data = modelling_data(lab)   # the same function train_within_1y.py and experiments.py use
frame = data.frame
print(f"\nFeature table built in {time.time() - t0:.0f}s: {len(frame):,} usable cases, {frame.shape[1]} columns "
      f"(vocabularies decided on {TRAIN_YEARS}: {len(data.case_types)} case types, {len(data.sections)} law+section pairs)")
display(frame[frame["filing_year"] == 2017].sample(4, random_state=2).T)
print("Takeaway: 12 inputs, all known on the day of filing; nothing from the case's later life.")
```

## B2 · The time split: train on the past, test on a later year

**What.** Cases filed in **2015** train the candidate models; cases filed in **2016** are used to
choose between them; the winner is refitted on 2015 + 2016 and tested **once** on cases filed in
**2017**, which no model ever saw. 2010–2014 filings are left out (Part A6).

**Why not a random split.** The model will be used on *future* cases. A random split would let the
model learn 2017's own conditions (how fast courts had become that year) and overstate how well it
forecasts. Testing on a later year is the honest test, and it exposes *drift* (Part B12).

```python
model_rows = frame[frame["within_1y"].notna() & frame["filing_year"].between(2015, 2017)]
fit_part = model_rows[model_rows["filing_year"] == 2015]       # train candidates
choose_part = model_rows[model_rows["filing_year"] == 2016]    # choose between them
final_train = model_rows[model_rows["filing_year"] <= 2016]    # refit the winner
test = model_rows[model_rows["filing_year"] == 2017]           # test once

split = pd.DataFrame({"role": ["train candidates", "choose the model", "refit the winner", "test once (unseen)"],
                      "filing years": ["2015", "2016", "2015-2016", "2017"],
                      "cases": [len(fit_part), len(choose_part), len(final_train), len(test)],
                      "share within a year": [p["within_1y"].mean() for p in (fit_part, choose_part, final_train, test)]})
display(split.set_index("role").style.format({"cases": "{:,}", "share within a year": "{:.1%}"}))
check("2017 test cases", len(test), R["test"]["baseline: overall rate"]["n"], tol=0.5)
print(f"Takeaway: {len(final_train):,} cases to learn from, {len(test):,} later cases to test on; "
      f"the share finishing within a year rises every year ({fit_part['within_1y'].mean():.0%} -> "
      f"{choose_part['within_1y'].mean():.0%} -> {test['within_1y'].mean():.0%}), which a random split would hide.")
```

## B3 · Baselines, and how we score a prediction

**What.** Three simple rules every model has to beat, each fitted on 2015–2016 and scored on 2017:

1. **Overall rate**: predict the same share for everyone.
2. **Rate per case type**: look up the share for that kind of case.
3. **Rate per case type × district**: a lookup table by kind of case and district (what a
   "college predictor"-style website would do). Small groups are shrunk towards the overall rate.

**How predictions are scored** (the model gives each case a *probability* of finishing within a year):

- **Accuracy**: call it "yes" when the probability is at least 50%; the share of cases where that
  call was right.
- **AUC**: take one case that did finish within a year and one that did not, at random; AUC is
  the chance the model gave the first one the higher probability. 0.5 = coin toss, 1.0 = perfect
  ranking.
- **Brier score**: the average of (probability − outcome)², the outcome being 1 or 0. Lower is
  better; always saying 50% scores 0.25.
- **Log loss**: also lower is better, but confident wrong answers (saying 95% when the case
  dragged on) are punished much harder. The model was *chosen* by log loss.

```python
# the project's own scoring functions, from scripts/train_within_1y.py
from train_within_1y import lookup, metrics

y_train, y_test = final_train["within_1y"].astype(int).to_numpy(), test["within_1y"].astype(int).to_numpy()
baselines = {
    "baseline: overall rate": np.full(len(test), final_train["within_1y"].mean()),
    "baseline: rate per case type": lookup(final_train, test, ["case_type"]),
    "baseline: rate per case type x district": lookup(final_train, test, ["case_type", "district"]),
}
scores = {name: metrics(y_test, p) for name, p in baselines.items()}
for name, s in scores.items():
    check(f"{name}: 2017 AUC", s["auc"], R["test"][name]["auc"])
display(pd.DataFrame(scores).T[["accuracy", "auc", "brier", "log_loss"]].style.format("{:.3f}"))
print(f"Takeaway: knowing only the kind of case already gives AUC {scores['baseline: rate per case type']['auc']:.3f}; "
      f"adding the district lifts it to {scores['baseline: rate per case type x district']['auc']:.3f}. "
      f"A model has to beat that lookup table to be worth anything.")
```

## B4 · Three model families, with and without the law inputs

**What.** `scripts/train_within_1y.py` trained, for each of the two input sets, three families
with a small grid of settings:

- **Logistic regression**: a weighted sum of the inputs (each category one-hot encoded, i.e. turned
  into its own 0/1 column; numbers standardised) squashed into a probability. Simple, fast, and its
  weights can be read.
- **Linear SVM**: finds the straight boundary that separates the two classes with the widest
  margin; its scores are turned into probabilities by a calibration step.
- **Random forest**: hundreds of decision trees, each grown on a random sample of cases and inputs,
  voting together. Can learn interactions on its own.

Each was fitted on 2015, scored on 2016, and the best setting of each family was refitted on
2015–2016 and scored on 2017. The forests take minutes; their results are read from
`reports/within_1y_results.json`. The logistic regression is fast, so the served configuration is
**refitted live here** and compared with the report.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from train_within_1y import one_hot

CATS, NUMS = FEATURE_SETS["+ law and section"]
t0 = time.time()
lr_live = Pipeline([("prep", one_hot(CATS, NUMS)), ("model", LogisticRegression(C=1.0, max_iter=3000))])
lr_live.fit(final_train[CATS + NUMS], y_train)
p_live = lr_live.predict_proba(test[CATS + NUMS])[:, 1]
live = metrics(y_test, p_live)
print(f"logistic regression (+ law and section, C=1.0) refitted on 2015-2016 in {time.time() - t0:.0f}s")
for m in ("auc", "brier", "log_loss", "accuracy"):
    check(f"live logistic regression 2017 {m}", live[m], R["test"][META["model"]][m])


def short(name):
    base = {"baseline: overall rate": "Baseline: overall rate",
            "baseline: rate per case type": "Baseline: rate per case type",
            "baseline: rate per case type x district": "Baseline: case type x district lookup"}
    if name in base:
        return base[name]
    inputs, rest = name.split(" | ")
    family = rest.split(" (")[0].replace(" (calibrated)", "").replace("linear SVM", "Linear SVM")
    family = family[0].upper() + family[1:]
    return f"{family}, {'filing basics' if inputs == 'filing basics' else 'with law and section'}"


comparison = pd.DataFrame(R["test"]).T[["accuracy", "auc", "brier", "log_loss"]].astype(float)
comparison.index = [short(n) + (" (served)" if n == META["model"] else "") for n in comparison.index]
comparison["source"] = ["computed above" if "Baseline" in n or "served" in n else "reports/within_1y_results.json"
                        for n in comparison.index]
display(comparison.style.format("{:.3f}", subset=["accuracy", "auc", "brier", "log_loss"])
        .highlight_max(subset=["accuracy", "auc"], color=ps.BLUE_RAMP[0])
        .highlight_min(subset=["brier", "log_loss"], color=ps.BLUE_RAMP[0]))

fig, ax = plt.subplots(figsize=(8.5, 4.6))
auc = comparison["auc"].drop("Baseline: overall rate")
ps.hbar(ax, list(auc.index), list(auc.values), fmt="{:.3f}", highlight={n for n in auc.index if "served" in n})
ax.set_xlim(0.5, 0.92)
ax.set_title("2017 AUC by model (bars start at 0.5, a coin toss; served model in orange)")
plt.show()

fig, ax = plt.subplots(figsize=(8.5, 4.6))
brier = comparison["brier"].drop("Baseline: overall rate")
ps.hbar(ax, list(brier.index), list(brier.values), fmt="{:.3f}", highlight={n for n in brier.index if "served" in n})
ax.set_title("2017 Brier score by model (lower is better; served model in orange)")
plt.show()

basics = comparison[comparison.index.str.contains("filing basics")]
with_law = comparison[comparison.index.str.contains("law and section")]
print(f"Takeaway: without the law inputs the best model reaches AUC {basics['auc'].max():.3f}, barely above the "
      f"lookup table ({comparison.loc['Baseline: case type x district lookup', 'auc']:.3f}); with them all three "
      f"families reach {with_law['auc'].min():.3f}-{with_law['auc'].max():.3f} and Brier about {with_law['brier'].mean():.3f}. "
      f"Knowing what the case is about is what helps; the choice of algorithm barely matters.")
```

## B5 · The served model, and why it was chosen

**What.** The rule, fixed *before* looking at 2017: serve whichever (input set, family, setting)
had the **lowest log loss on 2016** when trained on 2015. That was logistic regression with the law
inputs (C = 1.0, a regularisation setting; a smaller C pulls the weights towards zero).

**Why not the random forest**, which has a slightly higher 2017 AUC? Because the choice was made
on 2016, and picking a model by its test score would turn the test into part of training. The
logistic regression is also 6 kB, scores a case in microseconds, and its weights give exact
explanations ("this input added so much to the odds"), which the site shows.

The saved model (`artifacts/within_1y/model.joblib`) is loaded below and compared with the live
refit from B4: they should give identical predictions.

```python
selection = pd.DataFrame(R["selection"]).T[["log_loss", "auc", "brier"]].astype(float).sort_values("log_loss")
selection.index = [n.replace("baseline: ", "Baseline: ") for n in selection.index]
print("Model selection: trained on 2015, scored on 2016, best log loss first")
display(selection.head(8).style.format("{:.4f}"))
chosen = min(R["selection"], key=lambda n: R["selection"][n]["log_loss"])
print(f"Chosen (lowest 2016 log loss): {chosen}\nServed model in metadata.json: {META['model']}")

served = joblib.load(ARTIFACT / "model.joblib")
p_test = served.predict_proba(test[CATS + NUMS])[:, 1]
print(f"\nSaved model file: {(ARTIFACT / 'model.joblib').stat().st_size / 1e3:.0f} kB, scikit-learn {META['scikit_learn']}")
print(f"Largest difference between the saved model's and the live refit's 2017 probabilities: {np.abs(p_test - p_live).max():.2e}")
print(f"Takeaway: the served model is {short(META['model'])}, chosen on 2016 before 2017 was looked at; "
      f"the saved file reproduces the refit exactly.")
```

## B6 · Evaluation on 549,273 unseen 2017 cases

**What.** The served model's probabilities for every 2017 case, scored four ways (B3), plus:

- a **confusion matrix**: calling "yes" at 50%, how many cases were correctly called yes (true
  positives), wrongly called yes (false positives), wrongly called no (false negatives) and
  correctly called no (true negatives);
- a **calibration curve**: group cases by predicted probability (0–10%, 10–20%, …) and compare the
  average prediction in each group with the share that really finished within a year. Points on
  the diagonal mean "a predicted 70% really happens about 70% of the time".

## B7 · How much of that accuracy comes from easy cases?

**What.** The accuracy above is over *all* cases, and that hides something important. Petty summary
trials and bail-type petitions are almost always over within a year, so nearly anything gets them right.
This step scores the same predictions separately for those **easy cases** and for **real disputes**
(everything else), and compares each with the simplest possible rule: **always guess the commonest
answer for that group**.

**Why.** A model is only useful where it beats that rule. If we quoted only the overall accuracy we
would be overstating what filing-day information can do.

**Could a better model fix it?** We also trained logistic regression and a random forest on real disputes
only. If they did much better, the problem would be the model; if not, the information simply isn't
there on filing day.

```python
from tareekh.difficulty import difficulty_table

scored = test.assign(y=y_test, p=p_test)
table = difficulty_table(scored)
for group in ("all", "easy", "contested"):
    check(f"accuracy, {group} cases", table[group]["accuracy"], H["difficulty"][group]["accuracy"])

names = {"all": "All 2017 test cases", "easy": "Easy: summary trials and bail-type petitions", "contested": "Real disputes (everything else)"}
view = pd.DataFrame([{"group": names[g], "cases": table[g]["n"], "model accuracy": table[g]["accuracy"],
                      "always guess the commonest answer": table[g]["majority_accuracy"], "AUC": table[g]["auc"]} for g in names])
display(view.style.format({"cases": "{:,}", "model accuracy": "{:.1%}", "always guess the commonest answer": "{:.1%}", "AUC": "{:.3f}"}).hide(axis="index"))

types = pd.DataFrame(table["by_type"])
types["case type"] = types["case_type"].map(lambda t: CASE_TYPES.get(t, t.upper()).split(" (")[0])
big = types[types["n"] >= 9000].copy()
fig, ax = plt.subplots(figsize=(8.5, 4.2))
y = np.arange(len(big))
ax.barh(y + 0.19, big["accuracy"], height=0.36, color=ps.SERIES[0], label="model")
ax.barh(y - 0.19, big["majority_accuracy"], height=0.36, color=ps.NEUTRAL, label="always guess the commonest answer")
ax.set_yticks(y, big["case type"])
ax.xaxis.set_major_formatter(PercentFormatter(1.0))
ax.grid(axis="y", visible=False)
ax.invert_yaxis()
ax.legend(loc="lower right")
ax.set_title("Accuracy inside each kind of case, 2017")
plt.show()

only = pd.DataFrame(H["contested_only_models"]).T
display(only.style.format({"auc": "{:.3f}", "accuracy": "{:.1%}"}))

c = table["contested"]
print(f"Takeaway: the headline {table['all']['accuracy']:.1%} is mostly the model telling easy cases from hard ones. On real disputes "
      f"({c['n'] / table['all']['n']:.0%} of cases) it is right {c['accuracy']:.1%} of the time, against {c['majority_accuracy']:.1%} for always "
      f"guessing 'not within a year'. Models trained only on real disputes reach AUC {only['auc'].max():.2f}, so the limit is the "
      f"information available on filing day, not the algorithm.")
```

## B8 · What drives the predictions

**What.** Two views.

- **Permutation importance** (from the training script, on 50,000 test cases): shuffle one input's
  values across cases so it carries no information, and measure how much worse the log loss gets.
  A big increase means the model relied on that input.
- **The weights of the served model**: a logistic regression adds up one weight per category.
  A positive weight pushes towards "finishes within a year", a negative one towards "drags on".
  The largest ones are listed with their category names.

```python
imp = pd.Series({k: v["mean"] for k, v in R["served"]["permutation_importance_2017"].items()}).sort_values()
fig, ax = plt.subplots(figsize=(8, 4.2))
ps.hbar(ax, list(imp.index), list(imp.values), fmt="{:.3f}")
ax.set_title("Permutation importance: increase in log loss when one input is shuffled (2017)")
plt.show()

prep, lr = served.named_steps["prep"], served.named_steps["model"]
names = list(prep.get_feature_names_out())


def readable(name):
    name = name.replace("cat__", "").replace("num__", "")
    for prefix, label in [("case_type_", "case type"), ("law_section_", "section"), ("law_", "law"),
                          ("court_level_", "court"), ("district_", "district")]:
        if name.startswith(prefix):
            value = name[len(prefix):]
            if prefix == "case_type_" and value in CASE_TYPES:
                value = f"{value}: {CASE_TYPES[value].split(' (')[0].lower()}"
            return f"{label} = {value}"
    return name


weights = pd.Series(lr.coef_[0], index=[readable(n) for n in names])
top = pd.concat([weights.nlargest(8), weights.nsmallest(8)]).sort_values()
fig, ax = plt.subplots(figsize=(9.5, 5.2))
colors = [ps.SERIES[2] if w > 0 else ps.SERIES[1] for w in top.values]
bars = ax.barh(top.index, top.values, color=colors, height=0.7, edgecolor=ps.SURFACE, linewidth=2)
for bar, w in zip(bars, top.values, strict=True):
    ax.text(w + (0.08 if w > 0 else -0.08), bar.get_y() + bar.get_height() / 2, f"{w:+.2f}",
            va="center", ha="left" if w > 0 else "right", fontsize=8, color=ps.INK_2)
ax.axvline(0, color=ps.INK_2, lw=1)
ax.grid(axis="y", visible=False)
ax.set_xlim(top.min() * 1.3, top.max() * 1.3)
ax.set_title("Largest weights of the served model (green: faster; orange: slower)")
ax.set_xlabel("weight on the log-odds of finishing within a year")
plt.show()

print(f"Takeaway: the kind of case carries most of the signal (importance {imp['case_type']:.2f}, next "
      f"{imp.index[-2]} {imp.iloc[-2]:.2f}); every other input is below {imp.iloc[-3]:.2f}. "
      f"The model mostly knows which kinds of case are quick.")
```

## B9 · Side analysis: linear regression on the number of days

**What.** The main model answers yes/no because unfinished cases have no known duration. A
regression (predicting a number) needs finished cases, so `scripts/regress_days.py` asks a
narrower question and says so: **among cases disposed of within 730 days, how many days did they

log(1 + days) because durations are heavily skewed. The baseline is the median days for that kind

**Why it is exploratory only.** It describes cases that *did* finish within two years; it cannot
say anything about the many that did not, so it must not be read as "how long your case will take".

```python
from regress_days import HORIZON, report
from sklearn.linear_model import LinearRegression

finished = frame.join(lab[["observed_days", "ended"]])
finished = finished[finished["ended"] & (finished["observed_days"] <= HORIZON)]
reg_train, reg_test = finished[finished.filing_year == 2015], finished[finished.filing_year == 2016]
reg = Pipeline([("prep", one_hot(CATS, NUMS)), ("model", LinearRegression())])
reg.fit(reg_train[CATS + NUMS], np.log1p(reg_train["observed_days"]))
pred_days = np.expm1(reg.predict(reg_test[CATS + NUMS]))
base_days = reg_test["case_type"].map(reg_train.groupby("case_type")["observed_days"].median()).fillna(
    reg_train["observed_days"].median()).to_numpy()
days = reg_test["observed_days"].to_numpy()
reg_results = pd.DataFrame({"linear regression": report(days, pred_days),
                            "baseline: median per case type": report(days, base_days)}).T
reg_results.columns = ["test cases", "mean absolute error (days)", "median absolute error (days)", "R² on log(days)"]
display(reg_results.style.format({"test cases": "{:,.0f}", "mean absolute error (days)": "{:.0f}",
                                  "median absolute error (days)": "{:.0f}", "R² on log(days)": "{:.3f}"}))
check("regression mean absolute error (days)", reg_results.iloc[0, 1], D["linear_regression"]["mae_days"], tol=0.05)
check("regression R² on log(days)", reg_results.iloc[0, 3], D["linear_regression"]["r2_log_days"])

sample = reg_test.sample(6000, random_state=0).index
fig, ax = plt.subplots(figsize=(5.4, 5.2))
ax.scatter(days[reg_test.index.get_indexer(sample)] + 1, pred_days[reg_test.index.get_indexer(sample)] + 1,
           s=5, alpha=0.25, color=ps.SERIES[0], edgecolors="none")
ax.plot([1, HORIZON], [1, HORIZON], color=ps.NEUTRAL, lw=1, ls="--")
ax.set_xscale("log")
ax.set_yscale("log")
ax.set_xlabel("actual days + 1 (log scale)")
ax.set_ylabel("predicted days + 1 (log scale)")
ax.set_title("Days to disposal: predicted vs actual, 6,000 test cases filed in 2016")
plt.show()

print(f"Takeaway: among cases finished within two years the regression is typically off by "
      f"{reg_results.iloc[0, 2]:.0f} days (median error) and explains {reg_results.iloc[0, 3]:.0%} of the variation in "
      f"log(days), better than the per-type median ({reg_results.iloc[1, 3]:.0%}); its mean error "
      f"({reg_results.iloc[0, 1]:.0f} days) is worse than the baseline's ({reg_results.iloc[1, 1]:.0f}) because a few "
      f"long cases dominate the mean. Useful as a description, not as a forecast.")
```

## B10 · Side analysis: k-means on the districts

**What.** Each of the 33 districts gets a profile: its share of cases finished within a year for six
big kinds of case, plus the share of its 2015 filings still open at the cutoff. The profiles are
standardised (so no single measure dominates) and `clusters.choose_and_fit` runs **k-means** (which
groups points so each is close to its group's centre) for k = 2…8, choosing k by the **silhouette
score**: how much closer each district is to its own group than to the nearest other group, from
−1 to 1. Values above about 0.5 mean clear groups; below 0.25 the structure is weak.

**Why report it as weak.** The best silhouette is about 0.19. Districts differ by degree, not by
type; we show the groups as a summary and do not pretend to have discovered "types of district".

```python
from tareekh.clusters import choose_and_fit

profile_types = ["stc", "os", "cc", "mcop", "ep", "hmop"]
el["district"] = el["dist_code"].map(district_names)
profile = el[el["type"].isin(profile_types)].pivot_table(index="district", columns="type", values="within_1y", aggfunc="mean")
profile.columns = [f"within_1y_{c}" for c in profile.columns]
profile["open_at_cutoff_2015_filings"] = lab[lab["filing_year"] == 2015].assign(
    district=lambda d: d["dist_code"].map(district_names)).groupby("district")["ended"].apply(lambda s: 1 - s.mean())
clusters = choose_and_fit(profile)
check("best silhouette score", max(clusters.silhouette.values()), max(tn["district_clusters"]["silhouette"].values()))
print(f"chosen k = {clusters.k}; districts whose missing measure was filled with the median: {clusters.imputed}")

ks = sorted(clusters.silhouette)
fig, axes = plt.subplots(1, 2, figsize=(9, 3.4))
axes[0].plot(ks, [clusters.inertia[k] for k in ks], marker="o", color=ps.SERIES[0])
axes[0].set_title("Elbow: within-cluster sum of squares")
axes[1].plot(ks, [clusters.silhouette[k] for k in ks], marker="o", color=ps.SERIES[0])
axes[1].set_title("Silhouette score (higher = clearer groups)")
axes[1].set_ylim(0, 0.6)
axes[1].axhspan(0, 0.25, color=ps.GRID, alpha=0.6)
axes[1].text(2.1, 0.03, "weak structure", fontsize=8, color=ps.INK_2)
for ax in axes:
    ax.set_xlabel("k (number of groups)")
    ax.set_xticks(ks)
fig.tight_layout()
plt.show()

centres = clusters.centres.copy()
centres.index = [f"group {i}" for i in centres.index]
centres["districts"] = [", ".join(sorted(clusters.labels[clusters.labels == i].index)) for i in range(clusters.k)]
display(centres.style.format({c: "{:.0%}" for c in centres.columns if c != "districts"}))
print(f"Takeaway: k = {clusters.k} by silhouette, but the best silhouette is only "
      f"{max(clusters.silhouette.values()):.2f}: the groups are a convenient summary (faster, middling, slower "
      f"districts), not a discovery.")
```

## B11 · How long cases stay open: Kaplan–Meier curves

**What.** For Tamil Nadu filings 2015–2018, the share of cases **still open after N months**,
estimated with `survival.kaplan_meier`. At each day it multiplies together the survival of all
earlier days: S(t) = Π (1 − dᵢ/nᵢ), where dᵢ cases ended on day i out of nᵢ still being watched.
A pending case is watched until the cutoff and then simply leaves the count; a moved case leaves
at its move. Nothing is guessed about what happened after we stopped seeing a case.

**Assumption, stated.** Kaplan–Meier assumes that *why* we stopped watching is unrelated to how
soon the case would have ended. For the scrape cutoff that is true; for transfers it is an
approximation.

```python
from tareekh.survival import still_open_at

recent = lab[(lab["unusable"] == "") & lab["filing_year"].between(2015, 2018)]
show = [("all cases", recent)] + [(t, recent[recent["type"] == t]) for t in ("stc", "hmop", "os", "mcop", "cc")]
months = np.arange(0, 49)   # 2015 filings were watched for about 48 months before the cutoff
fig, ax = plt.subplots(figsize=(8.5, 4.8))
ends = []
table_rows = {}
for i, (name, part) in enumerate(show):
    km = kaplan_meier(part["observed_days"], part["ended"])
    vals = still_open_at(km, months * 30.44)
    label = "all cases" if name == "all cases" else CASE_TYPES[name].split(" (")[0]
    color = ps.INK_2 if name == "all cases" else ps.SERIES[i - 1]
    ax.plot(months, vals, color=color, lw=2.5 if name == "all cases" else 2, ls="--" if name == "all cases" else "-")
    ends.append([float(vals[-1]), label, color])
    table_rows[label] = {"cases": len(part), "median days": median_days(km),
                         **{f"open after {m} months": vals[m] for m in (12, 24, 48)}}
ends.sort()
for j in range(1, len(ends)):   # push the end labels apart so none overlap
    ends[j][0] = max(ends[j][0], ends[j - 1][0] + 0.055)
for y, text, color in ends:
    ax.annotate(text, (48, y), xytext=(5, 0), textcoords="offset points", va="center", fontsize=9, color=color)
ax.set_xlim(0, 48)
ax.set_ylim(0, 1)
ax.set_xticks([0, 12, 24, 36, 48])
ax.yaxis.set_major_formatter(PercentFormatter(1.0))
ax.set_xlabel("months since filing")
ax.set_title("Share of cases still open after N months (Tamil Nadu filings 2015-2018)")
plt.show()
curves = pd.DataFrame(table_rows).T
check("Kaplan-Meier median days, all cases", curves.loc["all cases", "median days"], tn["still_open_curves"]["all cases"]["median_days"], tol=0.5)
check("all cases still open after 12 months", curves.loc["all cases", "open after 12 months"], tn["still_open_curves"]["all cases"]["still_open"]["12"])
display(curves.style.format({"cases": "{:,.0f}", "median days": "{:,.0f}", "open after 12 months": "{:.0%}",
                             "open after 24 months": "{:.0%}", "open after 48 months": "{:.0%}"}))
print(f"Takeaway: half of all cases are over after {curves.loc['all cases', 'median days']:,.0f} days; "
      f"{curves.loc['all cases', 'open after 12 months']:.0%} are still open after a year and "
      f"{curves.loc['all cases', 'open after 48 months']:.0%} after four. Summary trials end in days "
      f"(median {curves.loc['Summary trial case', 'median days']:.0f}); calendar cases take years "
      f"(median {curves.loc['Calendar case', 'median days']:,.0f}).")
```

## B12 · Limitations, stated honestly

**What.** `scripts/experiments.py` probed the served model four ways; the results are read here.

1. **Within one kind of case it ranks poorly.** AUC inside a single case type is 0.49–0.62 for rent,
   domestic-violence, maintenance, execution, marriage and civil-suit cases. The model knows which
   *kinds* of case are quick; between two cases of the same kind it can barely tell.
2. **Drift.** Courts got faster every year, so a model trained on 2015–16 under-predicts 2017
   (mean prediction below the observed share). Weighting 2016 more barely helps.
3. **Other horizons.** The same recipe for "within 6 months" and "within 2 years" also beats the
   per-type baseline, so the finding is not specific to one year.
4. **Cutoff rule.** The stricter and looser scrape-cutoff rules give identical 2017 results.

Also: Tamil Nadu only; cases filed 2015–2017 and collected in early 2019, so courts have changed
since; the congestion inputs on the live site are the values at the end of the data; and it is a
statistical summary of past cases, not a prediction for any real case or legal advice.

```python
weak = pd.DataFrame(E["weak_spots"]["case_type"])
weak["kind of case"] = weak["case_type"].map(lambda t: CASE_TYPES.get(t, t).split(" (")[0])
print("1. AUC within a single kind of case (2017, kinds with 2,000+ cases; weakest first):")
display(weak.set_index("kind of case")[["auc", "n", "mean_predicted", "observed"]].head(8)
        .rename(columns={"auc": "AUC within the kind", "n": "cases", "mean_predicted": "mean prediction", "observed": "observed share"})
        .style.format({"AUC within the kind": "{:.2f}", "cases": "{:,}", "mean prediction": "{:.0%}", "observed share": "{:.0%}"}))

drift = pd.DataFrame(E["drift"]).T[["auc", "brier", "mean_predicted", "observed"]].astype(float)
print("2. Drift (test: 2017):")
display(drift.rename(columns={"auc": "AUC", "brier": "Brier", "mean_predicted": "mean prediction", "observed": "observed share"})
        .style.format({"AUC": "{:.3f}", "Brier": "{:.3f}", "mean prediction": "{:.1%}", "observed share": "{:.1%}"}))

horizons = pd.DataFrame({k: {"model AUC": v["model"]["auc"], "per-type baseline AUC": v["baseline: rate per case type"]["auc"],
                             "train years": str(v["train"]), "test year": v["test"]} for k, v in E["horizons"].items()}).T
print("3. Other horizons:")
display(horizons)

rule = {k: v["model_2017"]["auc"] for k, v in E["cutoff_rule"].items()}
print(f"4. Cutoff rule: 2017 AUC with the looser rule {rule['share 0.25']:.4f}, with the stricter rule {rule['share 0.5']:.4f}")

served_drift = drift.loc["train 2015-16 (served)"]
print(f"\nTakeaway: within one kind of case the AUC drops to {weak['auc'].min():.2f}-{weak['auc'].iloc[5]:.2f} for the six "
      f"weakest kinds; the model predicts {served_drift['mean_predicted']:.1%} for 2017 when {served_drift['observed']:.1%} "
      f"happened (courts kept getting faster). It is a statistical summary of 2015-2017 Tamil Nadu cases, not a forecast "
      f"for any real case.")
```

## Summary

**Part A.** 81 million raw eCourts rows, read as text; 29,857 law spellings, free-text courts and
sections, impossible dates and sentinel codes cleaned by small tested rules with every change
counted. Two facts decided the design: the data was collected in January–February 2019 (estimated
from the collapse in monthly decisions, per district), and the early years are missing their quick
cases, so only 2015–2017 filings are modelled. "Disposed of within a year" is labelled only for
cases that had the full year, which removes a bias that had made 2018 look over 90% "fast".

**Part B.** Twelve filing-day inputs, no leakage, no gender. Train 2015, choose on 2016, test once
on 2017. Three model families against three baselines: the law inputs matter far more than the
algorithm. The served logistic regression is right about 8 times in 10 on 549,273 unseen cases
(AUC 0.873, Brier 0.138 vs 0.844 / 0.163 for a lookup table) and its probabilities are calibrated.
